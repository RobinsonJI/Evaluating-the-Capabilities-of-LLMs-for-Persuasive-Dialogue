# Saving m2m debates

```python
model2model_debates_path = Path() / ".." / "persuasio" / "persuasio" / "outputs" / "dialogues"
```

```python
m2m_debates = []

for p in model2model_debates_path.rglob("*"):
    if p.is_file() and ("grok-4" not in str(p)) and ("gpt-5"):
        print(p)
        with open(p, "r", encoding="utf-8") as f:
            d = json.load(f)
            m2m_debates.append(d)

len(m2m_debates)
```

```python
models = ["gpt-4o", "grok-3", "mistral-medium"]
agent_variants = ["_base", "_mas", "_mas_rag"]

models_and_variants = [models[x] + agent_variants[y] for x in range(len(models)) for y in range(len(agent_variants))]

print(models_and_variants)

m2m_dialogue_count = 0
h2m_dialogue_count = 0
h2h_dialogue_count = 0

for idx, row in all_df.iterrows():
    if row["first_speaker"] in models_and_variants and row["second_speaker"] in models_and_variants:
        m2m_dialogue_count += 1

    elif row["first_speaker"] in models_and_variants and row["second_speaker"] not in models_and_variants:
        h2m_dialogue_count += 1

    elif row["first_speaker"] not in models_and_variants and row["second_speaker"] not in models_and_variants:  
        h2m_dialogue_count += 1
    else:
        h2h_dialogue_count += 1
```

```python
print(m2m_dialogue_count) 
print(h2m_dialogue_count) 
print(h2h_dialogue_count) 
```
