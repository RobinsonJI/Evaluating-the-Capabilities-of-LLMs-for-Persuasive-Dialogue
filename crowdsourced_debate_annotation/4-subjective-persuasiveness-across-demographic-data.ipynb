```python
df = pd.read_csv(
    "data/019c5684-e8e6-737c-9442-07692f856379.csv"
)

df = df[~df["Debate Topic"].str.contains("ATTENTION CHECK:", na=False)]
```

```python
df.head()
```

```python
comparison_ids = df["META_COMPARISONID"].unique().tolist()

annotator_cols = []
for col in df.columns:
    if col.startswith("Annotator") and col.endswith('_ID'):
        base = col.replace("_ID", "")
        annotator_cols.append((
            col,                          # ID column
            base + "_Timestamp",          # Timestamp column
            base + "_Response"            # Response column
        ))
annotator_cols
```

```python
def create_subgroup_df(main_df:pd.DataFrame, subgroup_df:pd.DataFrame, filename:str) -> pd.DataFrame:
    list_of_dict_data = []
    for idx, row in subgroup_df.iterrows():
        for col_id, col_ts, col_resp in annotator_cols:
            filtered = main_df[main_df[col_id] == row["Participant id"]]
            if not filtered.empty:
                for f_idx, f_row in filtered.iterrows():
                    for f_col_id, f_col_ts, f_col_resp in annotator_cols:
                        if main_df.at[f_idx, f_col_id] == row["Participant id"]:

                            data = {
                                    "DataPoint_ID": f_row["DataPoint_ID"],
                                    "META_COMPARISONID": f_row["META_COMPARISONID"],
                                    "META_COMPARISON": f_row["META_COMPARISON"],
                                    "Question": f_row["Question"],
                                    "Annotator_ID": f_row[f_col_id],
                                    "Annotator_Response": f_row[f_col_resp],
                                }
                            list_of_dict_data.append(data)

    combined_df = pd.DataFrame(list_of_dict_data)

    combined_df.to_csv(f"data/{filename}.csv")
```

```python
new_df = create_subgroup_df(
    main_df=df,
    subgroup_df=left_pol_aff_df,
    filename="self-declared-left-leaning-annotators"
)
```
